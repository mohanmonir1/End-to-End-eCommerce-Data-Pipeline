## Bronze control table

```sql
CREATE TABLE IF NOT EXISTS bronze.control.ingestion_control (
    run_id STRING,
    table_id BIGINT,
    last_successful_ts TIMESTAMP,
    last_successful_pk BIGINT,
    rows_written BIGINT,
    run_status STRING,
    updated_at TIMESTAMP
  ) USING DELTA;

ALTER TABLE
  bronze.control.ingestion_control
ALTER COLUMN
  run_id
  SET NOT NULL;

ALTER TABLE
  bronze.control.ingestion_control
ALTER COLUMN
  table_id
  SET NOT NULL;

ALTER TABLE
  bronze.control.ingestion_control
DROP CONSTRAINT IF EXISTS
  pk_ingestion_control;

ALTER TABLE
  bronze.control.ingestion_control
ADD
  CONSTRAINT pk_ingestion_control PRIMARY KEY (run_id, table_id);

CREATE TABLE IF NOT EXISTS bronze.control.object_list (
    table_id BIGINT,
    table_name STRING,
    pk_col STRING,
    ts_col STRING,
    layer STRING
  ) USING DELTA;

ALTER TABLE
  bronze.control.object_list
ALTER COLUMN
  table_id
  SET NOT NULL;

ALTER TABLE
  bronze.control.object_list
DROP CONSTRAINT IF EXISTS
  pk_object_list;

ALTER TABLE
  bronze.control.object_list
ADD
  CONSTRAINT pk_object_list PRIMARY KEY (table_id);
```

## Silver Control Table

This table stores the latest Silver processing state for each entity.

It helps us track:
- the latest Bronze run already processed by Silver
- the latest Bronze ingestion timestamp already processed
- how many rows were merged in the latest Silver run

```sql
CREATE TABLE IF NOT EXISTS silver.control.object_list (
    table_id BIGINT,
    table_name STRING,
    layer STRING,
    join_key STRING
  ) USING DELTA;

ALTER TABLE
  silver.control.object_list
ALTER COLUMN
  table_id
  SET NOT NULL;

ALTER TABLE
  silver.control.object_list
DROP CONSTRAINT IF EXISTS
  pk_object_list;

ALTER TABLE
  silver.control.object_list
ADD
  CONSTRAINT pk_object_list PRIMARY KEY (table_id);

CREATE TABLE IF NOT EXISTS silver.control.processing_control (
    run_id STRING,
    table_id BIGINT,
    last_processed_at TIMESTAMP,
    rows_merged BIGINT,
    status STRING,
    updated_at TIMESTAMP
  ) USING DELTA;

ALTER TABLE
  silver.control.processing_control
ALTER COLUMN
  run_id
  SET NOT NULL;

ALTER TABLE
  silver.control.processing_control
ALTER COLUMN
  table_id
  SET NOT NULL;

ALTER TABLE
  silver.control.processing_control
DROP CONSTRAINT IF EXISTS
  pk_processing_control;

ALTER TABLE
  silver.control.processing_control
ADD
  CONSTRAINT pk_processing_control PRIMARY KEY (run_id, table_id);
```

```sql
INSERT INTO silver.control.object_list
  VALUES
    (1, "orders", "bronze", "order_id"),
    (2, "products", "bronze", "product_id"),
    (3, "payments", "bronze", "payment_id")
```

```sql
SELECT * FROM silver.control.object_list
```
