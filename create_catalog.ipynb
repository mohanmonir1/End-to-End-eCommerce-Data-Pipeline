## Create Bronze Catalog and Schema

```sql
%sql
CREATE CATALOG IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS bronze.raw;
CREATE SCHEMA IF NOT EXISTS bronze.control;
```

## Create Silver Catalog and Schema

```sql
%sql
CREATE CATALOG IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS silver.cleaned;
CREATE SCHEMA IF NOT EXISTS silver.transformed;
CREATE SCHEMA IF NOT EXISTS silver.control;
CREATE SCHEMA IF NOT EXISTS silver.bad_records;
```

## Create Gold Catalog and Schema

```sql
%sql
CREATE CATALOG IF NOT EXISTS gold;
CREATE SCHEMA IF NOT EXISTS gold.control;
CREATE SCHEMA IF NOT EXISTS gold.curated;
CREATE VOLUME IF NOT EXISTS gold.curated.snapshots;
```
